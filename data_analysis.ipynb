# Integrating SLLM-Generated Feedback

```python
import json

gemma_feedback_data = json.load(open("./gemma_4_generated_data.json"))
llama_8_feedback_data = json.load(open("./llama_8_generated_data.json"))
llama_70_feedback_data = json.load(open("./llama_70_generated_data.json"))
qwen_72_feedback_data = json.load(open("./qwen_72_generated_data.json"))
llama_1_feedback_data = json.load(open("./llama_1_generated_data.json"))
llama_3_feedback_data = json.load(open("./llama_3_generated_data.json"))
qwen_1_feedback_data = json.load(open("./qwen_1_generated_data.json"))
qwen_3_feedback_data = json.load(open("./qwen_3_generated_data.json"))
qwen_7_feedback_data = json.load(open("./qwen_7_generated_data.json"))
```

```python
for llama_70, llame_8, qwen_72, gemma, llama_1, llama_3, qwen_1, qwen_3, qwen_7 in zip(llama_70_feedback_data, 
                                                                                       llama_8_feedback_data, 
                                                                                       qwen_72_feedback_data, 
                                                                                       gemma_feedback_data,
                                                                                       llama_1_feedback_data,
                                                                                       llama_3_feedback_data,
                                                                                       qwen_1_feedback_data,
                                                                                       qwen_3_feedback_data,
                                                                                       qwen_7_feedback_data):
    passage_id = llama_70[0]
    llama_70_info = llama_70[1]
    llama_8_info = llame_8[1]
    qwen_72_info = qwen_72[1]
    gemma_info = gemma[1]
    llama_1_info = llama_1[1]
    llama_3_info = llama_3[1]
    qwen_1_info = qwen_1[1]
    qwen_3_info = qwen_3[1]
    qwen_7_info = qwen_7[1]
    
    for llama_70_history, llama_8_history, qwen_72_history, gemma_history, llama_1_history, llama_3_history, qwen_1_history, qwen_3_history, qwen_7_history in zip(llama_70_info,
                                                                 llama_8_info,
                                                                 qwen_72_info,
                                                                 gemma_info,
                                                                 llama_1_info,
                                                                 llama_3_info,
                                                                 qwen_1_info,
                                                                 qwen_3_info,
                                                                 qwen_7_info):
        generated_feedback = []
        for reply_candidate in llama_70_history["reply_candidates"]:
            if "gpt4" in reply_candidate:
                generated_feedback.append([0, "gpt4", reply_candidate[2]])
            elif "gpt3" in reply_candidate:
                generated_feedback.append([1, "gpt3", reply_candidate[2]])
        llama_70_feedback = llama_70_history["llama_70_feedback"]
        llama_8_feedback = llama_8_history["llama_8_feedback"]
        qwen_72_feedback = qwen_72_history["qwen_72_feedback"]
        gemma_feedback = gemma_history["gemma_4_feedback"]
        llama_1_feedback = llama_1_history["llama_1_feedback"]
        llama_3_feedback = llama_3_history["llama_3_feedback"]
        qwen_1_feedback = qwen_1_history["qwen_1_feedback"]
        qwen_3_feedback = qwen_3_history["qwen_3_feedback"]
        qwen_7_feedback = qwen_7_history["qwen_7_feedback"]
        generated_feedback.append([2, "llama_70", llama_70_feedback])
        generated_feedback.append([3, "llama_8", llama_8_feedback])
        generated_feedback.append([4, "qwen_72", qwen_72_feedback])
        generated_feedback.append([5, "gemma", gemma_feedback])
        generated_feedback.append([6, "llama_1", llama_1_feedback])
        generated_feedback.append([7, "llama_3", llama_3_feedback])
        generated_feedback.append([8, "qwen_1", qwen_1_feedback])
        generated_feedback.append([9, "qwen_3", qwen_3_feedback])
        generated_feedback.append([10, "qwen_7", qwen_7_feedback])
        llama_70_history["generated_feedback"] = generated_feedback  
    
```

```python
with open("./data/sllm_augmented_train_data.json", "w") as f:
    json.dump(llama_70_feedback_data, f, indent=4)
```
